# So sánh lỗi

```python
import pandas as pd

base_df = pd.read_csv("outputs/base_model_predictions.csv")
ft_df = pd.read_csv("outputs/finetuned_model_predictions.csv")

compare_df = base_df[["text", "label_name", "pred_label", "is_correct"]].copy()
compare_df = compare_df.rename(columns={
    "pred_label": "base_pred",
    "is_correct": "base_correct"
})

compare_df["ft_pred"] = ft_df["pred_label"]
compare_df["ft_correct"] = ft_df["is_correct"]

improved_df = compare_df[(compare_df["base_correct"] == False) & (compare_df["ft_correct"] == True)]
improved_df.to_csv("outputs/improved_cases.csv", index=False)

print("Saved improved cases to outputs/improved_cases.csv")
print("Num improved:", len(improved_df))
```
